```python
import torch

# Carregar arquivo .pt (PyTorch)
dl = torch.load('ds_test.pt', weights_only=False)

# Exibir o conteúdo carregado
for data, targets in dl:
    print(data.shape, targets.shape)
```

```python
# Arquivos de saída
data_file = 'dataset_data.txt'
targets_file = 'dataset_targets.txt'

with open(data_file, 'w') as fd, open(targets_file, 'w') as ft:
    for data, targets in dl:
      
        
        # Achatando os tensores caso sejam multidimensionais
        d_flat = data.flatten().numpy()
        t_flat = targets.flatten().numpy()
        
        # Convertendo os valores para string separados por espaço
        d_str = ' '.join(map(str, d_flat))
        t_str = ' '.join(map(str, t_flat))
        
        # Escrevendo no arquivo (uma linha por amostra)
        fd.write(d_str + '\n')
        ft.write(t_str + '\n')

print(f"Dataset salvo com sucesso: '{data_file}' e '{targets_file}'")
```
