# NeuroHls

```python
from neuro_hls import *
```

```python
def get_acc_from_vitis_log():
    import re

    with open("teste_quant_NMNIST/vitis_hls.log", "r", encoding="utf-8") as f:
        vitis_log = f.read()

    m = re.search(r"\*\*\* Final Acc: (.*?)%", vitis_log)

    return float(m.group(1))
```

```python
neuro_hls = NeuroHls("teste_quant_NMNIST")
nir_file = "nir_examples/cnn_sinabs.nir"
```

```python
quant_to_analyse = [(16, 8), (8, 4), (4, 2)]

for quant in quant_to_analyse:

    model = neuro_hls.read_nir_file(nir_file)

    model.input_quantization = quant
    model.weight_quantization = quant
    model.potential_quantization = quant

    neuro_hls.implement_model(model, use_float=False)

    neuro_hls.define_test_dataset("nir_examples/cnn_teste.npz", data_is_binary=True, step_count=300, different_sample_per_step=True)
    neuro_hls.create_testbench(total_samples=1024, batch_size=2, reset_potentials=True, debug_mode=False)

    neuro_hls.run_csim()

    acc = get_acc_from_vitis_log()

    with open("resultados_testes_quant_NMNIST/quant_to_acc_NMNIST.txt", "a", encoding="utf-8") as f:
        f.write(f"Quantization: {quant}  ->  Acc: {acc}\n")
```
